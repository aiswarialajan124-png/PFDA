# Assingment 03: Email Domains Pie Chart

```python
# Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load the data
URL = "https://drive.google.com/uc?id=1AWPf-pJodJKeHsARQK_RHiNsE8fjPCVK&export=download"

df = pd.read_csv(URL)
print(len(df))
df.head()
```

```python
# Extract the domain from each email
df["Domain"] = df["Email"].str.split("@").str[1]

df[["Email", "Domain"]].head()
```

```python
# Count the domains
domain_counts = df["Domain"].value_counts()
print(len(domain_counts), "different domains")
domain_counts
```

```python
# Draw piechart
colours = ["#4E79A7", "#F28E2B", "#59A14F"] # one colour per slice

fig, ax = plt.subplots(figsize=(8, 6))

wedges, texts, autotexts = ax.pie(
    domain_counts,
    labels=domain_counts.index,
    colors=colours,
    explode=[0.05, 0, 0],
    autopct="%1.1f%%", #show percentages
    startangle=90,     # first slice starts at the top
    pctdistance=0.75,  # percentages sit inside the slices 
    wedgeprops={"edgecolor": "white", "linewidth": 2}, # white lines between slices
    textprops={"fontsize": 12},
)
# Make the percentages white and bold so they stand out
for text in autotexts:
    text.set_color("white")
    text.set_fontweight("bold")

ax.set_title("Email Domains of 1000 people", fontsize=16, fontweight="bold")
ax.axis("equal")  # keeps the pie circular

plt.show()
```
